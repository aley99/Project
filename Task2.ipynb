```python
import pandas as pd

# Load the dataset
url = "https://raw.githubusercontent.com/Deepubhatt/EDA-on-Netflix-Dataset/main/netflix_dataset.csv"
df = pd.read_csv(url)

# Check for duplicate rows
duplicate_rows = df[df.duplicated()]
print(f"Number of duplicate rows: {duplicate_rows.shape[0]}")
```

```python
# Remove duplicate rows
df_cleaned = df.drop_duplicates()

# Verify removal
print(f"Number of rows after removing duplicates: {df_cleaned.shape[0]}")
```

```python
# Calculate the number of records lost
records_lost = df.shape[0] - df_cleaned.shape[0]
print(f"Number of records lost after cleaning: {records_lost}")
```
