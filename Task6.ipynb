```python
# Step 1: Import libraries
import pandas as pd

# Step 2: Load dataset from GitHub
url = "https://raw.githubusercontent.com/fivethirtyeight/uber-tlc-foil-response/master/uber-trip-data/uber-raw-data-apr14.csv"
df = pd.read_csv(url)

# Step 3: View the first few rows
print(df.head())

# Step 4: Convert Date/Time column to datetime format
df['Date/Time'] = pd.to_datetime(df['Date/Time'])

# Step 5: Create new features
df['hour_of_day'] = df['Date/Time'].dt.hour
df['day_of_week'] = df['Date/Time'].dt.day_name()

# Step 6: Verify new columns
print(df[['Date/Time', 'hour_of_day', 'day_of_week']].head())

# Step 7: Save cleaned dataset
df.to_csv("uber_pickups_feature_engineered.csv", index=False)
print("✅ Feature-engineered dataset saved as 'uber_pickups_feature_engineered.csv'")

```
