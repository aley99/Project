```python
# Import necessary library
import pandas as pd

# Load dataset from GitHub
url = "https://raw.githubusercontent.com/justmarkham/pandas-videos/master/data/imdb_1000.csv"
df = pd.read_csv(url)

# Show first few rows before cleaning
print("Before Cleaning:")
print(df.head())

# ---- STRING CLEANING & STANDARDIZATION ----

# 1. Remove extra spaces
df['title'] = df['title'].str.strip()
df['genre'] = df['genre'].str.strip()

# 2. Standardize capitalization
df['title'] = df['title'].str.title()
df['genre'] = df['genre'].str.lower()

# 3. Correct common typos (example: 'dramaa' -> 'drama')
df['genre'] = df['genre'].str.replace('dramaa', 'drama', regex=False)

# ---- Display cleaned data ----
print("\nAfter Cleaning:")
print(df.head())

# Optional: Check for null or invalid values
print("\nMissing values:\n", df.isnull().sum())
```
