```python
import pandas as pd

# URL to raw CSV in repo (you’ll need to get the “raw” link from GitHub)
url = "https://raw.githubusercontent.com/asharvi1/UCI-Air-Quality-Data/master/AirQualityUCI.csv"

# The dataset uses semicolon (“;”) as delimiter
df = pd.read_csv(url, sep=';')

# Show first few rows
print(df.head())

# Combine date and time columns into one datetime
df['Datetime'] = pd.to_datetime(
    df['Date'] + ' ' + df['Time'], 
    format='%d/%m/%Y %H.%M.%S', 
    errors='coerce'
)

# Extract useful features
df['Year'] = df['Datetime'].dt.year
df['Month'] = df['Datetime'].dt.month
df['Weekday'] = df['Datetime'].dt.day_name()

# Check result
print(df[['Datetime', 'Year', 'Month', 'Weekday']].head())

# Save cleaned version, if needed
df.to_csv("airquality_with_datetime_features.csv", index=False)
```
