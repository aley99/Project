```python
import pandas as pd

# Load dataset from GitHub
url = "https://raw.githubusercontent.com/jakobgross/Students-Performance-in-Exams-Kaggle-/master/StudentsPerformance.csv"
df = pd.read_csv(url)

# One-hot encode categorical columns
df_encoded = pd.get_dummies(df, columns=['gender', 'lunch', 'test preparation course'], drop_first=True)

# Display the first few rows of the encoded dataframe
print(df_encoded.head())

# Save the encoded dataframe to a new CSV file
df_encoded.to_csv("students_performance_encoded.csv", index=False)
```
