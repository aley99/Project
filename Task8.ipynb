```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Load dataset from GitHub
url = "https://gist.githubusercontent.com/pravalliyaram/5c05f43d2351249927b8a3f3cc3e5ecf/raw/Mall_Customers.csv"
df = pd.read_csv(url)

# Initialize MinMaxScaler
scaler = MinMaxScaler()

# Normalize 'Annual Income (k$)' and 'Spending Score (1–100)'
df[['Annual Income (k$)', 'Spending Score (1-100)']] = scaler.fit_transform(
    df[['Annual Income (k$)', 'Spending Score (1-100)']]
)

# Display the first few rows of the normalized dataframe
print(df.head())

# Save the normalized dataframe to a new CSV file
df.to_csv("mall_customers_normalized.csv", index=False)
```
