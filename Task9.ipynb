```python
import pandas as pd

# Load dataset from GitHub
url = "https://raw.githubusercontent.com/datasets/population/master/data/population.csv"
df = pd.read_csv(url)

# Display initial data types
print("Initial Data Types:")
print(df.dtypes)

# Convert 'Value' column to numeric, coercing errors to NaN
df['Value'] = pd.to_numeric(df['Value'], errors='coerce')

# Check for invalid entries (NaN values)
invalid_entries = df[df['Value'].isna()]
print("\nInvalid Entries (NaN values):")
print(invalid_entries)

# Display final data types
print("\nFinal Data Types:")
print(df.dtypes)

# Save the cleaned dataset to a new CSV file
df.to_csv("world_population_cleaned.csv", index=False)
```
